```python
import os
import random
from PIL import Image

for train_val in ['train', 'val']:
    for annotation in ['l_cheek', 'r_cheek', 'chin', 'forehead', 'glabellus', 'l_peroucular', 'r_peroucular']:

        # 원본 이미지 폴더 경로
        folder_paths = [
            f'/content/drive/MyDrive/Final_project_2조/02_2. 전처리 및 EDA_이미지/data/image/Orientation/{train_val}/{annotation}/smart_pad',
            f'/content/drive/MyDrive/Final_project_2조/02_2. 전처리 및 EDA_이미지/data/image/Orientation/{train_val}/{annotation}/{annotation}_origin'
        ]

        # 저장 경로
        save_path = f'/content/drive/MyDrive/Final_project_2조/02_2. 전처리 및 EDA_이미지/data/image/Orientation/{train_val}/{annotation}/rotation_plus_10'

        # 저장 경로가 존재하지 않으면 생성
        os.makedirs(save_path, exist_ok=True)

        all_files = []

        # 두 폴더에서 파일 목록을 모두 가져오기
        for folder_path in folder_paths:
            all_files.extend([os.path.join(folder_path, f) for f in os.listdir(folder_path) if f.endswith('.jpg')])

        # 전체 파일 중 30%를 랜덤으로 추출
        selected_files = random.sample(all_files, int(0.3 * len(all_files)))

        for img_path in selected_files:
            filename = os.path.basename(img_path)
            # 이미지 열기
            img = Image.open(img_path)

            # 이미지 10도 회전
            rotated_img = img.rotate(10, expand=True)

            # 저장 경로 생성
            save_img_path = os.path.join(save_path, filename)

            # 회전된 이미지 저장
            rotated_img.save(save_img_path)

        print(f"30%의 이미지를 350도 회전하여 저장")
```

```python
import cv2
import numpy as np
import os
import random

for train_val in ['train', 'val']:

    # 원본 이미지 폴더 경로
    folder_paths = [
        f'/content/drive/MyDrive/Final_project_2조/02_2. 전처리 및 EDA_이미지/data/image/Orientation/{train_val}/chin/smart_pad',
        f'/content/drive/MyDrive/Final_project_2조/02_2. 전처리 및 EDA_이미지/data/image/Orientation/{train_val}/chin/chin_origin'
    ]
    background_image_path = '/content/drive/MyDrive/Final_project_2조/02_2. 전처리 및 EDA_이미지/지우님/흰색.png'

    # 저장 경로
    save_path = f'/content/drive/MyDrive/Final_project_2조/02_2. 전처리 및 EDA_이미지/data/image/Orientation/{train_val}/chin/background'

    # 저장 경로가 존재하지 않으면 생성
    os.makedirs(save_path, exist_ok=True)

    all_files = []

    # 두 폴더에서 파일 목록을 모두 가져오기
    for folder_path in folder_paths:
        all_files.extend([os.path.join(folder_path, f) for f in os.listdir(folder_path) if f.endswith('.jpg')])

    # 전체 파일 중 30%를 랜덤으로 추출
    selected_files = random.sample(all_files, int(0.3 * len(all_files)))

    for img_path in selected_files:
        # 이미지 불러오기
        image = cv2.imread(img_path)

        # 이미지 파일이 제대로 읽혔는지 확인
        if image is None:
            print(f"이미지를 읽는 데 실패했습니다: {img_path}")
            continue

        # 객체 분리 (간단한 예로 OpenCV의 grabCut 사용)
        mask = np.zeros(image.shape[:2], np.uint8)
        bgd_model = np.zeros((1, 65), np.float64)
        fgd_model = np.zeros((1, 65), np.float64)

        # 관심 영역(rect) 설정
        rect = (10, 10, image.shape[1]-20, image.shape[0]-20)
        cv2.grabCut(image, mask, rect, bgd_model, fgd_model, 5, cv2.GC_INIT_WITH_RECT)

        # 마스크 적용하여 객체(턱 부분) 분리
        mask2 = np.where((mask == 2) | (mask == 0), 0, 1).astype('uint8')
        image_foreground = image * mask2[:, :, np.newaxis]

        # 배경 이미지 불러오기 및 크기 조정
        background = cv2.imread(background_image_path)
        if background is None:
            print(f"배경 이미지를 읽는 데 실패했습니다: {background_image_path}")
            continue

        background = cv2.resize(background, (image.shape[1], image.shape[0]))

        # 배경과 객체 합성
        mask2_inv = cv2.bitwise_not(mask2)
        background_cut = background * mask2_inv[:, :, np.newaxis]
        result = background_cut + image_foreground

        # 저장 경로 생성
        filename = os.path.basename(img_path) # 파일이름만 추출
        save_img_path = os.path.join(save_path, filename)   # 저장할 전체 경로

        # 이미지 저장
        cv2.imwrite(save_img_path, result)    # 이미지 저장할 때 사용

    print(f'{save_path}에 저장')
```
